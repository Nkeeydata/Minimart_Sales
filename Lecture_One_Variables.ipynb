## Introduction

Variables are fundamental to programming—they act as containers for storing data values. In Python, variables are created when you assign a value to them. Unlike some other programming languages, Python does not require explicit declaration of variable types; the type is inferred from the value assigned.

## Variable Naming Rules

When naming variables in Python, follow these rules:
- Variable names must start with a letter (a–z, A–Z) or an underscore (_).
- The rest of the variable name can include letters, numbers (0–9), and underscores.
- Variable names are case-sensitive (`age`, `Age`, and `AGE` are different variables).
- Avoid using Python reserved keywords (e.g., `if`, `for`, `while`, `class`, `def`, etc.).

## Examples of Valid and Invalid Variable Names

| Variable Name | Valid/Invalid | Reason |
|---------------|---------------|--------|
| `age`         | Valid         | Follows all naming rules. |
| `2nd_name`    | Invalid       | Starts with a number. |
| `first_name`  | Valid         | Uses letters and underscore. |
| `first-name`  | Invalid       | Hyphens are not allowed. |
| `class`       | Invalid       | `class` is a reserved keyword. |
